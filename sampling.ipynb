```python
import pandas as pd

file_path = "/synfs/notebook/1/aml_notebook_mount/Users/smustudent08/NF-UQ-NIDS-v2.csv"
```

```python
df = pd.read_csv(file_path, nrows=5)

print(df.head())
print(df.columns)
```

```python
drop_columns = [
    'DNS_QUERY_ID',
    'DNS_QUERY_TYPE',
    'DNS_TTL_ANSWER',
    'FTP_COMMAND_RET_CODE',
    'ICMP_TYPE',
    'ICMP_IPV4_TYPE'
]
```

```python
target_classes = [
    'Benign',
    'scanning',
    'Reconnaissance',
    'xss',
    'password',
    'injection'
]
```

```python
chunks = []

for chunk in pd.read_csv(
    file_path,
    chunksize=100000
):

    # 컬럼 제거
    chunk = chunk.drop(columns=drop_columns)

    # 필요한 공격만 선택
    chunk = chunk[
        chunk['Attack'].isin(target_classes)
    ]

    chunks.append(chunk)

df = pd.concat(chunks)

print(df['Attack'].value_counts())
```

```python
print(df.shape)
```

```python
sample_sizes = {
    'Benign': 1500000,
    'scanning': 600000,
    'Reconnaissance': 600000,
    'xss': 600000,
    'password': 600000,
    'injection': 600000
}
```

```python
sampled_dfs = []

sampled_indices = []
```

```python
for attack_type, size in sample_sizes.items():

    sampled = (
        df[df['Attack'] == attack_type]
        .sample(n=size, random_state=42)
    )

    sampled_dfs.append(sampled)

    sampled_indices.extend(sampled.index)

    print(f"{attack_type} 샘플링 완료")
```

```python
final_df = pd.concat(sampled_dfs)

final_df = final_df.sample(
    frac=1,
    random_state=42
)
#학습용데이터 생성
```

```python
print(final_df['Attack'].value_counts())

print(final_df.shape)
```

```python
final_df.to_csv(
    "sampled_dataset.csv",
    index=False
)
#학습용 데이터 저장
```
